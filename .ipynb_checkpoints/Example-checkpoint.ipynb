# If Statement

if <condition>:
  <code>
elif <condition>:
  <code>
...

  <code>

```python
# do not forget the space in if statment and else if
```

```python
age = 15
```

```python
if age >= 18:
    print("You're an adult! ")
elif age>=13:
    print("You're a teenager")
else:
    print("You're a kid")
```

# For loop

for <variable> in <list>:
  <code>

```python
countries
```

```python
for country in countries:
    print(country)
```

```python
#enumerate() is a built-in Python function that adds an automatic counter to a collection, returning each item alongside its index as a tuple (e.g., (0, 'japan')).
```

```python
for country in enumerate (countries):
    print(country)
```

```python
for i ,country in enumerate (countries):
    print(i)
    print(country)
```

# Function

def function (<params>):
    <code>
    return <data>

In Python, a function is a reusable block of code defined with the def keyword that runs only when you call it. It allows you to pass data in as arguments, perform a specific task, and return a result, which keeps your code organized and prevents repetition.

```python
def sum_values (a , b):
    x=a + b
    return x
```

```python
# 5 = a , 8 = b
sum_values(5 , 8)
```
