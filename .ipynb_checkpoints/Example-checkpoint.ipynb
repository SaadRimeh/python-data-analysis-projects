# Function

def function (<params>):
    <code>
    return <data>

In Python, a function is a reusable block of code defined with the def keyword that runs only when you call it. It allows you to pass data in as arguments, perform a specific task, and return a result, which keeps your code organized and prevents repetition.

```python
def sum_values (a , b):
    x=a + b
    return x
```

```python
# 5 = a , 8 = b
sum_values(5 , 8)
```

# Module

In Python, a module is simply a file containing Python code (like functions, variables, or classes) that you can bring into your current project using the import keyword, allowing you to organize and reuse your code easily.

import

## OS Module

```python
import os
```

os.getcwd() stands for "Get Current Working Directory." It is a function from the os module that returns the exact file path of the folder where your Python script is currently running.

```python
os.getcwd()
```

os.listdir() is a function from the os module that returns a list of all the file and folder names inside a specific directory. If you leave the parentheses empty, it shows everything in your current working directory.

```python
os.listdir()
```

os.makedirs("New Folder") is a function from the os module that creates a new directory (folder) named "New Folder" in your current working directory. It is especially powerful because it can create multiple nested folders all at once if you give it a longer path (like os.makedirs("project/data/images")).

```python
os.makedirs("New Folder")
```

```python
os.listdir()
```

# Creating a dataframe from an array

## Optin 1

```python
import pandas as pd
import numpy as np
```

```python
# creating an array
data = np.array([[1, 4] ,[2 , 5],[3,6]])
```

```python
data
```

This code uses the pandas library to create a structured data table (called a DataFrame) named df using existing data, while assigning custom names to its three rows (index) and two columns (columns).

```python
# creating a dataframe
df=pd.DataFrame(data , index=['row1' ,'row2' , 'row3'] , 
            columns=['col1' , 'col2'])
```

```python
df
```
